```python

# ============================================================
# SMART PARKING PROJECT: CNN vs RNN vs LSTM
# ONE CELL - GOOGLE COLAB
# ============================================================

import os, sys, random, tarfile, shutil, time, zipfile, subprocess
import xml.etree.ElementTree as ET
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

# ---------------- 1. INSTALL ----------------
print("Installing libraries...")
subprocess.check_call([
    sys.executable, "-m", "pip", "install", "-q",
    "huggingface_hub", "opencv-python-headless",
    "scikit-learn", "seaborn", "tqdm"
])

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from tqdm.auto import tqdm
from huggingface_hub import hf_hub_download
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)
from IPython.display import display

# ---------------- 2. SETTINGS ----------------
SEED = 42
IMG_SIZE = 64
BATCH_SIZE = 32
EPOCHS = 6

# Small sample for faster execution
FRAMES = {"train": 60, "val": 20, "test": 20}

# Maximum crops PER CLASS
LIMITS = {"train": 800, "val": 200, "test": 200}

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

BASE = Path("/content/Smart_Parking_Final")
RAW = Path("/content/PKLot_Raw")
DATA = BASE / "dataset"
RESULTS = BASE / "results"
MODELS = BASE / "models"
PREVIEWS = BASE / "parking_lot_outputs"

for folder in [BASE, RAW, DATA, RESULTS, MODELS, PREVIEWS]:
    folder.mkdir(parents=True, exist_ok=True)

print("TensorFlow:", tf.__version__)
print("GPU available:", bool(tf.config.list_physical_devices("GPU")))

CLASSES = ["Occupied", "Vacant"]

# ---------------- 3. REUSE OR DOWNLOAD DATASET ----------------
print("\n[1/10] Checking PKLot dataset...")

xml_files = list(RAW.rglob("*.xml")) if RAW.exists() else []

if len(xml_files) < 100:
    print("Dataset not extracted. Checking download cache...")
    archive = hf_hub_download(
        repo_id="teenygrad/pklot",
        repo_type="dataset",
        filename="PKLot.tar.gz"
    )

    print("Extracting PKLot archive...")
    with tarfile.open(archive, "r:gz") as tar:
        try:
            tar.extractall(path=RAW, filter="data")
        except TypeError:
            tar.extractall(path=RAW)

    xml_files = list(RAW.rglob("*.xml"))

if not xml_files:
    raise RuntimeError("PKLot XML annotations were not found.")

print("XML annotation files found:", len(xml_files))

# ---------------- 4. MATCH XML WITH ORIGINAL IMAGES ----------------
print("\n[2/10] Finding annotated parking-lot images...")

pairs = []

for x in xml_files:
    candidates = [
        x.with_suffix(".jpg"),
        x.with_suffix(".JPG"),
        x.with_suffix(".jpeg"),
        x.with_suffix(".JPEG")
    ]

    img = next((p for p in candidates if p.exists()), None)

    if img is not None:
        pairs.append((img, x))

print("Image/XML pairs:", len(pairs))

if len(pairs) < 100:
    raise RuntimeError("Too few matching image/XML pairs found.")

# ---------------- 5. SPLIT BY ACQUISITION DAY ----------------
print("\n[3/10] Creating train, validation and test splits...")

groups = {}

for img, xml in pairs:
    key = str(xml.parent.relative_to(RAW))
    groups.setdefault(key, []).append((img, xml))

group_names = list(groups.keys())
random.shuffle(group_names)

n = len(group_names)
a = int(n * 0.70)
b = int(n * 0.85)

group_splits = {
    "train": group_names[:a],
    "val": group_names[a:b],
    "test": group_names[b:]
}

selected = {}

for split, names in group_splits.items():
    candidates = [
        item for name in names for item in groups[name]
    ]
    random.shuffle(candidates)
    selected[split] = candidates[:FRAMES[split]]
    print(split, "frames selected:", len(selected[split]))

# ---------------- 6. EXTRACT PARKING SPACE CROPS ----------------
print("\n[4/10] Extracting a small sample of parking spaces...")

reservoir = {
    (split, cls): []
    for split in ["train", "val", "test"]
    for cls in CLASSES
}

seen = {
    (split, cls): 0
    for split in ["train", "val", "test"]
    for cls in CLASSES
}

# Keep a few original test frames for full-lot visual output
preview_frames = []

def tag(element):
    return element.tag.split("}")[-1].lower()

def get_contour(space):
    for child in space.iter():
        if tag(child) == "contour":
            return child
    return None

def crop_parking_space(image, points):
    try:
        pts = np.array(points, dtype=np.float32)
        if len(pts) < 4:
            return None

        # Four-corner approximation of the parking-space polygon
        rect = cv2.boxPoints(cv2.minAreaRect(pts))
        sums = rect.sum(axis=1)
        diffs = np.diff(rect, axis=1).reshape(-1)

        src = np.array([
            rect[np.argmin(sums)],
            rect[np.argmin(diffs)],
            rect[np.argmax(sums)],
            rect[np.argmax(diffs)]
        ], dtype=np.float32)

        dst = np.array([
            [0, 0], [IMG_SIZE-1, 0],
            [IMG_SIZE-1, IMG_SIZE-1], [0, IMG_SIZE-1]
        ], dtype=np.float32)

        matrix = cv2.getPerspectiveTransform(src, dst)
        return cv2.warpPerspective(image, matrix, (IMG_SIZE, IMG_SIZE))

    except Exception:
        return None

def add_sample(split, cls, crop):
    key = (split, cls)
    seen[key] += 1

    ok, encoded = cv2.imencode(".jpg", crop)
    if not ok:
        return

    data = encoded.tobytes()
    bucket = reservoir[key]
    limit = LIMITS[split]

    if len(bucket) < limit:
        bucket.append(data)
    else:
        j = random.randint(0, seen[key]-1)
        if j < limit:
            bucket[j] = data

for split in ["train", "val", "test"]:

    for img_path, xml_path in tqdm(
        selected[split], desc=f"Extracting {split}"
    ):
        image = cv2.imread(str(img_path))
        if image is None:
            continue

        try:
            root = ET.parse(str(xml_path)).getroot()
        except Exception:
            continue

        preview_polygons = []
        preview_crops = []

        for space in root.iter():
            if tag(space) != "space":
                continue

            occupied = space.attrib.get("occupied")
            if occupied not in ["0", "1"]:
                continue

            contour = get_contour(space)
            if contour is None:
                continue

            points = []
            for point in contour.iter():
                if tag(point) == "point":
                    try:
                        points.append([
                            float(point.attrib["x"]),
                            float(point.attrib["y"])
                        ])
                    except Exception:
                        pass

            if len(points) < 4:
                continue

            crop = crop_parking_space(image, points)
            if crop is None:
                continue

            cls = "Occupied" if occupied == "1" else "Vacant"
            add_sample(split, cls, crop)

            if split == "test" and len(preview_frames) < 3:
                preview_polygons.append(points)
                preview_crops.append(crop)

        if split == "test" and len(preview_frames) < 3:
            preview_frames.append({
                "image": image.copy(),
                "polygons": preview_polygons,
                "crops": preview_crops
            })

# ---------------- 7. SAVE BALANCED DATA ----------------
print("\n[5/10] Saving CNN/RNN/LSTM datasets...")

for split in ["train", "val", "test"]:
    for cls in CLASSES:
        folder = DATA / split / cls
        folder.mkdir(parents=True, exist_ok=True)

        for i, content in enumerate(reservoir[(split, cls)]):
            (folder / f"space_{i:05d}.jpg").write_bytes(content)

        print(
            split, cls,
            ":", len(reservoir[(split, cls)])
        )

for split in ["train", "val", "test"]:
    for cls in CLASSES:
        if len(reservoir[(split, cls)]) < 30:
            raise RuntimeError(
                f"Not enough {cls} examples in {split}. "
                "Try increasing the number of selected frames."
            )

del reservoir

# ---------------- 8. DATASET DISTRIBUTION ----------------
print("\n[6/10] Dataset distribution...")

dist = []

for split in ["train", "val", "test"]:
    for cls in CLASSES:
        count = len(list((DATA / split / cls).glob("*.jpg")))
        dist.append({
            "Split": split,
            "Class": cls,
            "Images": count
        })

dist_df = pd.DataFrame(dist)
display(dist_df)

plt.figure(figsize=(8, 5))
sns.barplot(data=dist_df, x="Split", y="Images", hue="Class")
plt.title("PKLot Sample Distribution")
plt.tight_layout()
plt.savefig(RESULTS / "dataset_distribution.png", dpi=300)
plt.show()

# ---------------- 9. LOAD TENSORFLOW DATA ----------------
print("\n[7/10] Loading image datasets...")

train_ds = tf.keras.utils.image_dataset_from_directory(
    str(DATA / "train"),
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_names=CLASSES,
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    str(DATA / "val"),
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_names=CLASSES,
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    str(DATA / "test"),
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_names=CLASSES,
    shuffle=False
)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

# ---------------- 10. BUILD THREE MODELS ----------------
print("\n[8/10] Building CNN, RNN and LSTM...")

def build_cnn():
    return tf.keras.Sequential([
        tf.keras.layers.Input((IMG_SIZE, IMG_SIZE, 3)),
        tf.keras.layers.Rescaling(1./255),
        tf.keras.layers.RandomFlip("horizontal"),
        tf.keras.layers.Conv2D(32, 3, activation="relu", padding="same"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.Conv2D(64, 3, activation="relu", padding="same"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.Conv2D(128, 3, activation="relu", padding="same"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.GlobalAveragePooling2D(),
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(2, activation="softmax")
    ])

def build_rnn():
    return tf.keras.Sequential([
        tf.keras.layers.Input((IMG_SIZE, IMG_SIZE, 3)),
        tf.keras.layers.Rescaling(1./255),
        tf.keras.layers.Reshape((IMG_SIZE, IMG_SIZE * 3)),
        tf.keras.layers.SimpleRNN(48),
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(2, activation="softmax")
    ])

def build_lstm():
    return tf.keras.Sequential([
        tf.keras.layers.Input((IMG_SIZE, IMG_SIZE, 3)),
        tf.keras.layers.Rescaling(1./255),
        tf.keras.layers.Reshape((IMG_SIZE, IMG_SIZE * 3)),
        tf.keras.layers.LSTM(48),
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(2, activation="softmax")
    ])

builders = {
    "CNN": build_cnn,
    "RNN": build_rnn,
    "LSTM": build_lstm
}

# ---------------- 11. TRAIN AND EVALUATE ----------------
print("\n[9/10] Training all three models...")

y_true = np.concatenate([y.numpy() for x, y in test_ds])

histories = {}
models = {}
predictions = {}
metrics = {}

for name, builder in builders.items():

    print("\n" + "="*45)
    print("TRAINING", name)
    print("="*45)

    tf.keras.backend.clear_session()
    model = builder()

    model.compile(
        optimizer=tf.keras.optimizers.Adam(0.001),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    callbacks = [
        tf.keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=2,
            restore_best_weights=True
        ),
        tf.keras.callbacks.ModelCheckpoint(
            str(MODELS / f"best_{name.lower()}.keras"),
            monitor="val_loss",
            save_best_only=True
        )
    ]

    start = time.time()

    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=EPOCHS,
        callbacks=callbacks,
        verbose=1
    )

    elapsed = time.time() - start

    loss, acc = model.evaluate(test_ds, verbose=0)
    probs = model.predict(test_ds, verbose=0)
    pred = np.argmax(probs, axis=1)

    histories[name] = history
    models[name] = model
    predictions[name] = pred

    metrics[name] = {
        "Accuracy": accuracy_score(y_true, pred),
        "Precision": precision_score(y_true, pred, average="weighted", zero_division=0),
        "Recall": recall_score(y_true, pred, average="weighted", zero_division=0),
        "F1 Score": f1_score(y_true, pred, average="weighted", zero_division=0),
        "Test Loss": loss,
        "Training Time (sec)": elapsed
    }

    model.save(MODELS / f"{name.lower()}_final.keras")

# ---------------- 12. COMPARISON GRAPHS ----------------
print("\n[10/10] Creating final comparison and parking outputs...")

# Accuracy and loss curves
for metric, title, filename in [
    ("accuracy", "Training Accuracy", "training_accuracy.png"),
    ("loss", "Training Loss", "training_loss.png")
]:
    plt.figure(figsize=(10, 5))

    for name, history in histories.items():
        plt.plot(history.history[metric], label=f"{name} train")
        val_metric = "val_" + metric
        plt.plot(
            history.history[val_metric],
            linestyle="--",
            label=f"{name} validation"
        )

    plt.title(title + " - CNN vs RNN vs LSTM")
    plt.xlabel("Epoch")
    plt.ylabel(metric.capitalize())
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(RESULTS / filename, dpi=300)
    plt.show()

# Confusion matrices and reports
for name in builders:
    cm = confusion_matrix(y_true, predictions[name], labels=[0, 1])

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm, annot=True, fmt="d", cmap="Blues",
        xticklabels=CLASSES, yticklabels=CLASSES
    )
    plt.title(name + " Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.savefig(RESULTS / f"{name.lower()}_confusion_matrix.png", dpi=300)
    plt.show()

    report = classification_report(
        y_true, predictions[name],
        labels=[0, 1], target_names=CLASSES,
        zero_division=0
    )

    print("\n", name, "classification report\n", report)

    (RESULTS / f"{name.lower()}_report.txt").write_text(report)

# Model comparison
comparison = pd.DataFrame(metrics).T
comparison.index.name = "Model"
comparison = comparison.sort_values("F1 Score", ascending=False)

display(comparison)
comparison.to_csv(RESULTS / "model_comparison.csv")

best_model = comparison.index[0]
print("\nBEST MODEL:", best_model)
print("Selected using weighted F1-score.")

for col, filename in [
    ("Accuracy", "accuracy_comparison.png"),
    ("Precision", "precision_comparison.png"),
    ("Recall", "recall_comparison.png"),
    ("F1 Score", "f1_comparison.png")
]:
    plt.figure(figsize=(8, 5))
    plt.bar(comparison.index, comparison[col] * 100)
    plt.title("CNN vs RNN vs LSTM - " + col)
    plt.ylabel(col + " (%)")
    plt.ylim(0, 100)
    plt.tight_layout()
    plt.savefig(RESULTS / filename, dpi=300)
    plt.show()

# ---------------- CNN PARKING AVAILABILITY ----------------
print("\nCNN PARKING AVAILABILITY")

cnn_pred = predictions["CNN"]
vacant = int(np.sum(cnn_pred == 1))
occupied = int(np.sum(cnn_pred == 0))
total = len(cnn_pred)

availability = 100 * vacant / total if total else 0

print("Test parking spaces:", total)
print("Vacant / available:", vacant)
print("Occupied:", occupied)
print("Availability:", round(availability, 2), "%")

pd.DataFrame({
    "Status": ["Vacant / Available", "Occupied"],
    "Count": [vacant, occupied]
}).to_csv(RESULTS / "parking_availability.csv", index=False)

plt.figure(figsize=(7, 6))
plt.pie(
    [vacant, occupied],
    labels=["Vacant / Available", "Occupied"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("CNN Parking Availability")
plt.tight_layout()
plt.savefig(RESULTS / "parking_availability.png", dpi=300)
plt.show()

# ---------------- FULL PARKING-LOT VISUAL OUTPUT ----------------
# CNN predicts each annotated parking space and marks it:
# Green = Vacant; Red = Occupied

print("\nCreating CNN parking-lot marked images...")

for frame_no, item in enumerate(preview_frames, 1):

    image = item["image"].copy()
    crops = item["crops"]
    polygons = item["polygons"]

    if not crops:
        continue

    batch = np.asarray(crops, dtype=np.float32)
    probabilities = models["CNN"].predict(batch, verbose=0)
    pred_labels = np.argmax(probabilities, axis=1)

    frame_vacant = 0
    frame_occupied = 0

    for points, label in zip(polygons, pred_labels):
        polygon = np.array(points, dtype=np.int32).reshape((-1, 1, 2))

        if label == 1:
            color = (0, 220, 0)
            text = "V"
            frame_vacant += 1
        else:
            color = (0, 0, 255)
            text = "O"
            frame_occupied += 1

        cv2.polylines(image, [polygon], True, color, 2)

        center = np.mean(np.array(points), axis=0).astype(int)
        cv2.putText(
            image, text, tuple(center),
            cv2.FONT_HERSHEY_SIMPLEX, 0.45, color, 1
        )

    cv2.imwrite(
        str(PREVIEWS / f"parking_lot_{frame_no}.jpg"),
        image
    )

    print(
        f"Frame {frame_no}: vacant={frame_vacant}, "
        f"occupied={frame_occupied}"
    )

    display(
        __import__("PIL").Image.open(
            PREVIEWS / f"parking_lot_{frame_no}.jpg"
        )
    )

# ---------------- SAVE PROJECT SUMMARY ----------------
summary = f"""
SMART PARKING SPACE CLASSIFICATION

Dataset: PKLot
Models compared: CNN, RNN, LSTM

BEST MODEL BY WEIGHTED F1 SCORE: {best_model}

CNN:
{metrics['CNN']}

RNN:
{metrics['RNN']}

LSTM:
{metrics['LSTM']}

CNN TEST PARKING AVAILABILITY:
Total test spaces: {total}
Vacant: {vacant}
Occupied: {occupied}
Vacancy percentage: {availability:.2f}%

Vacant means unoccupied in the dataset.
It does not guarantee physical or personal safety.
"""

(RESULTS / "project_summary.txt").write_text(summary)
print(summary)

# ---------------- ZIP ALL OUTPUTS ----------------
zip_path = "/content/Smart_Parking_CNN_RNN_LSTM.zip"

print("Creating ZIP file...")

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for file in BASE.rglob("*"):
        if file.is_file():
            z.write(file, file.relative_to(BASE))

print("\nPROJECT COMPLETED!")
print("Models:", MODELS)
print("Graphs and reports:", RESULTS)
print("Parking-lot output images:", PREVIEWS)
print("ZIP:", zip_path)

from google.colab import files
files.download(zip_path)
```
